Hangman game :)

```python
import random

# Large word list
words = [
    "python", "hangman", "programming", "computer", "science", "mathematics",
    "algorithm", "function", "variable", "iteration", "condition", "loop",
    "dictionary", "framework", "development", "software", "hardware",
    "internet", "database", "security", "network", "artificial",
    "intelligence", "machine", "learning", "neural", "analysis",
    "statistics", "geometry", "algebra", "calculus", "physics",
    "chemistry", "biology", "astronomy", "philosophy", "psychology",
    "engineering", "architecture", "education", "knowledge", "language",
    "history", "government", "economics", "finance", "entrepreneur",
    "innovation", "technology", "communication", "strategy", "design",
    "creativity", "leadership", "management", "productivity", "success",
    "challenge", "opportunity", "adventure", "experience", "discovery",
    "exploration", "imagination", "inspiration", "motivation", "determination",
    "discipline", "confidence", "achievement", "excellence", "progress"
]

high_score = 0

def choose_word(difficulty):
    if difficulty == "easy":
        filtered = [w for w in words if len(w) <= 6]
    elif difficulty == "medium":
        filtered = [w for w in words if 7 <= len(w) <= 9]
    else:  # hard
        filtered = [w for w in words if len(w) >= 10]

    return random.choice(filtered)

def play_game():
    global high_score

    print("\nSelect difficulty:")
    print("1 - Easy")
    print("2 - Medium")
    print("3 - Hard")

    choice = input("Enter choice (1/2/3): ")

    if choice == "1":
        difficulty = "easy"
        multiplier = 1
    elif choice == "2":
        difficulty = "medium"
        multiplier = 2
    elif choice == "3":
        difficulty = "hard"
        multiplier = 3
    else:
        print("Invalid choice. Defaulting to Easy.")
        difficulty = "easy"
        multiplier = 1

    word = choose_word(difficulty)
    guessed_letters = []
    tries = 6

    print(f"\n🎮 Starting {difficulty.upper()} mode!")
    print("You have 6 tries.")

    while tries > 0:
        display_word = ""
        for letter in word:
            if letter in guessed_letters:
                display_word += letter + " "
            else:
                display_word += "_ "

        print("\nWord:", display_word.strip())
        print("Tries left:", tries)
        print("Guessed letters:", ", ".join(guessed_letters))

        guess = input("Enter a letter: ").lower()

        if len(guess) != 1 or not guess.isalpha():
            print("⚠️ Enter a single valid letter.")
            continue

        if guess in guessed_letters:
            print("⚠️ Already guessed.")
            continue

        guessed_letters.append(guess)

        if guess in word:
            print("✅ Correct!")
        else:
            print("❌ Wrong!")
            tries -= 1

        if all(letter in guessed_letters for letter in word):
            print("\n🎉 You guessed the word:", word)
            score = tries * 10 * multiplier
            print("🏆 Score this round:", score)

            if score > high_score:
                high_score = score
                print("🔥 New High Score!")

            print("⭐ High Score:", high_score)
            return

    print("\n💀 You lost! The word was:", word)
    print("⭐ High Score:", high_score)


# Game loop
while True:
    play_game()
    again = input("\nPlay again? (yes/no): ").lower()
    if again != "yes":
        print("Thanks for playing! Final High Score:", high_score)
        break
```

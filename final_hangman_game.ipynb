```python
import random

# Word dictionary with categories (for hints)
word_bank = {
    "python": "Programming language",
    "hangman": "Classic word guessing game",
    "computer": "Electronic machine",
    "science": "Field of study",
    "algorithm": "Step-by-step problem solving method",
    "database": "Organized data storage",
    "network": "Connected systems",
    "physics": "Study of matter and energy",
    "biology": "Study of life",
    "chemistry": "Study of substances",
    "astronomy": "Study of space",
    "philosophy": "Study of fundamental questions",
    "architecture": "Building design field",
    "economics": "Study of money and markets",
    "innovation": "New idea creation",
    "technology": "Application of scientific knowledge",
    "leadership": "Guiding others",
    "management": "Organizing resources",
    "creativity": "Ability to create new ideas",
    "determination": "Strong will to succeed",
    "intelligence": "Ability to learn and reason",
    "engineering": "Applied science field",
    "mathematics": "Study of numbers",
    "communication": "Exchange of information",
    "development": "Process of growth"
}

words = list(word_bank.keys())

high_score = 0

def choose_word(difficulty):
    if difficulty == "easy":
        filtered = [w for w in words if len(w) <= 6]
    elif difficulty == "medium":
        filtered = [w for w in words if 7 <= len(w) <= 9]
    else:
        filtered = [w for w in words if len(w) >= 10]
    return random.choice(filtered)

def play_game():
    global high_score

    print("\nSelect difficulty:")
    print("1 - Easy")
    print("2 - Medium")
    print("3 - Hard")

    choice = input("Enter choice (1/2/3): ")

    if choice == "1":
        difficulty = "easy"
        multiplier = 1
    elif choice == "2":
        difficulty = "medium"
        multiplier = 2
    elif choice == "3":
        difficulty = "hard"
        multiplier = 3
    else:
        print("Invalid choice. Defaulting to Easy.")
        difficulty = "easy"
        multiplier = 1

    word = choose_word(difficulty)
    guessed_letters = []
    tries = 6
    hint_used = False

    print(f"\n🎮 {difficulty.upper()} MODE")
    print("Category hint:", word_bank[word])
    print("You have 6 tries.")
    print("Type 'hint' to reveal a letter (costs 1 try, once per game).")

    while tries > 0:
        display_word = ""
        for letter in word:
            if letter in guessed_letters:
                display_word += letter + " "
            else:
                display_word += "_ "

        print("\nWord:", display_word.strip())
        print("Tries left:", tries)
        print("Guessed letters:", ", ".join(guessed_letters))

        guess = input("Enter a letter: ").lower()

        if guess == "hint":
            if not hint_used:
                hidden_letters = [l for l in word if l not in guessed_letters]
                if hidden_letters:
                    revealed = random.choice(hidden_letters)
                    guessed_letters.append(revealed)
                    tries -= 1
                    hint_used = True
                    print(f"💡 Hint used! Revealed letter: {revealed}")
                else:
                    print("No letters left to reveal!")
            else:
                print("⚠️ You already used your hint.")
            continue

        if len(guess) != 1 or not guess.isalpha():
            print("⚠️ Enter a single valid letter.")
            continue

        if guess in guessed_letters:
            print("⚠️ Already guessed.")
            continue

        guessed_letters.append(guess)

        if guess in word:
            print("✅ Correct!")
        else:
            print("❌ Wrong!")
            tries -= 1

        if all(letter in guessed_letters for letter in word):
            print("\n🎉 You guessed the word:", word)
            score = tries * 10 * multiplier
            print("🏆 Score this round:", score)

            if score > high_score:
                high_score = score
                print("🔥 New High Score!")

            print("⭐ High Score:", high_score)
            return

    print("\n💀 You lost! The word was:", word)
    print("⭐ High Score:", high_score)


while True:
    play_game()
    again = input("\nPlay again? (yes/no): ").lower()
    if again != "yes":
        print("Thanks for playing! Final High Score:", high_score)
        break
```
